```python
import pandas as pd
df= pd.read_csv('../data/customers_shopping_data_enhanced.csv')
print(df.head())
```

```python
print(df.shape)
print(df.columns.tolist())
print(df.dtypes)
```

```python
missing=df.isnull().sum()
print(missing[missing>0])

```

```python
print(df.isnull().sum()/len(df)*100)
```

```python
print("duplicated rows:", df.duplicated().sum())
```

```python
df.drop_duplicates()
```

```python
df['invoice_date'].head()
```

```python
df['invoice_date']=pd.to_datetime(df['invoice_date'], format='%d/%m/%Y')
df['year']=df['invoice_date'].dt.year
df['month']=df['invoice_date'].dt.month
df['month_name']=df['invoice_date'].dt.strftime('%B')
df['day']=df['invoice_date'].dt.day
```

```python
df['gender']=df['gender'].str.strip().str.title()
df['category']=df['category'].str.strip().str.title()
df['customer_segment']=df['customer_segment'].str.strip().str.title()
```

```python
df['customer_rating']=df['customer_rating'].fillna(df['customer_rating'].median())
df['delivery_days']=df['delivery_days'].fillna(df['delivery_days'].median())
df['return_reason']=df['return_reason'].fillna('No Return')
```

```python
df.to_csv('../data/customers_shopping_data_cleaned.csv',index=False)
print("cleaned file saved")
df.shape
```
